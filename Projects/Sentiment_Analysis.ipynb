```python
from transformers import pipeline, AutoTokenizer , AutoModelForSequenceClassification
import torch
```

```python
classifier = pipeline(
    "text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)
```

```python
texts = [
    "I love this movie",
    "This is the worst film ever",
    "The story was okay but acting was great"
]
```

```python
results = classifier(texts)
```

```python
print("Pretrain Model Prediction : ")
for text, result in zip(texts, results):
  print(text)
  print(result)
  print()
```

```python
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-uncased")
model = AutoModelForSequenceClassification.from_pretrained("distilbert-base-uncased"
, num_labels=2)
```

```python
train_texts = ["I love this movie", "I hate this movue"]
labels = [1, 0]
```

```python
inputs = tokenizer(train_texts, padding=True, truncation=True, return_tensors="pt")
```

```python
outputs = model(**inputs, labels=torch.tensor(labels))
loss = outputs.loss
```

```python
print("Loss : ",loss.item())
```
