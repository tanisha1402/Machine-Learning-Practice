Importing the libraries


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# loading the dataset to a Pandas DataFrame
dataset = pd.read_csv('/content/Placement_Dataset.csv')
```

```python
dataset.head()
```

```python
dataset.shape
```

```python
dataset.isnull().sum()
```

Central Tendencies:
1. Mean -> Average
2. Median -> Sort the data in ascending order, then pick the middle value
3. Mode -> Pick the value with most frequency

```python
# analyaze the distribution of data in salary
fig , ax = plt.subplots(figsize=(8,6))
sns.distplot(dataset.salary)
```

```python
# since this is graph , where we are seeing a lot of data is in a particular place, and most other isn't . we shouldn't be using mean. Since , there isn't much of a even distribution going on berween data
# so, we can use median or mode
```

Replace the missing values with median

```python
dataset['salary'].fillna(dataset['salary'].median(),inplace=True)
```
