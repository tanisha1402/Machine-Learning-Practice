```python
import numpy as np
import cv2
from google.colab.patches import cv2_imshow
```

```python
img = cv2.imread('/content/scenario.jpg',0)
```

```python
cv2_imshow(img)
```

```python
cv2.imwrite('gray.jpg',img)
```

**How to resize image**

```python
img.shape
```

```python
resized_image = cv2.resize(img,(200,200))
```

```python
cv2_imshow(resized_image)
```

```python
crop=resized_image[0:200,20:]
```

```python
cv2_imshow(crop)
```

```python
rot = cv2.rotate(crop,cv2.ROTATE_90_CLOCKWISE)
```

```python
cv2_imshow(rot)
```

```python
canvas=np.zeros((300,300,3),dtype='uint8')
```

```python
cv2_imshow(canvas)
```

```python
img_line = canvas.copy()
```

```python
cv2.line(img_line,(50,50),(200,50),(255,0,0),2)
```

```python
img_rect=canvas.copy()
cv2.rectangle(img_rect,(50,50),(200,200),(0,255,0),2)
```

```python
cv2.circle(img_rect,(150,150),50,(0,0,255),2)
```

```python
cv2_imshow(img_rect)
```
